```python
import pandas as pd 
import os
```

```python
df = pd.read_csv(r'E:\VirtualWorks-Intership\task1-data-cleaning\data\netflix_titles.csv')
```

```python
print("Original Data Shape:", df.shape)
```

```python
print("\nMissing values before cleaning:")
print(df.isnull().sum())
```

```python
original_rows = len(df)
original_missing = int(df.isnull().sum().sum())
original_duplicates = int(df.duplicated().sum())
```

```python
df = df.drop_duplicates()
```

```python
df = df.replace(r'^\s+|\s+$', '', regex=True)
```

```python
columns_to_fill = [
    "director", "cast", "country",
    "date_added", "rating", "duration"
]
```

```python
for column in columns_to_fill:
    df[column] = df[column].fillna("Unknown")
```

```python
print("\nMissing values after cleaning:")
print(df.isnull().sum())
```

```python
print("\nDuplicate rows after cleaning:", df.duplicated().sum())
print("Cleaned dataset shape:", df.shape)
```

```python
os.makedirs("output", exist_ok=True)
df.to_csv("output/netflix_cleaned.csv", index=False)
```

```python
report = pd.DataFrame({
    "Metric": [
        "Original rows",
        "Cleaned rows",
        "Duplicate rows removed",
        "Missing values before cleaning",
        "Missing values after cleaning"
    ],
    "Value": [
        original_rows,
        len(df),
        original_duplicates,
        original_missing,
        int(df.isnull().sum().sum())
    ]
})
```

```python
report.to_csv("output/data_quality_report.csv", index=False)

print("\nCleaning completed successfully!")
print("Files saved in the output folder.")
```
