```python
import pandas as pd 
import os
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"E:\VirtualWorks-Intership\task2-customer-chur-analysis\data\WA_Fn-UseC_-Telco-Customer-Churn.csv")
```

```python
print(df.head())
print("Dataset shape:", df.shape)
```

```python
print("\nColumn names:")
print(df.columns)
```

```python
print("\nChurn distribution:")
print(df["Churn"].value_counts())
```

```python
print("\nMissing values:")
print(df.isnull().sum())
```

```python
print(df["Churn"].value_counts())
```

```python
print("Churn rate:", round((df["Churn"].eq("Yes").mean()) * 100, 2), "%")
```

```python
df["Churn"].value_counts().plot(kind="bar")
plt.title("Customer Churn Analysis")
plt.xlabel("Churn")
plt.ylabel("Number of Customers")
plt.show()
```

```python
print(df.groupby("Churn")["MonthlyCharges"].mean())
```

```python
df["Churn"].value_counts().to_csv("../output/churn_summary.csv")
```

```python
df.groupby("Churn")["MonthlyCharges"].mean().to_csv(
    "../output/monthly_charges.csv"
)
```
